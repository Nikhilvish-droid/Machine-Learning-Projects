```python
from pathlib import Path 
import pandas as pd

data_dir = Path("../data")

mon = pd.read_csv(data_dir/'Monday.csv')
tue = pd.read_csv(data_dir/'Tuesday.csv')
wed = pd.read_csv(data_dir/'Wednesday.csv')
th_m = pd.read_csv(data_dir/'Thursday-m.csv')
th_af = pd.read_csv(data_dir/'Thursday-aft.csv')
fri_m = pd.read_csv(data_dir/'Friday-m.csv')
fri_af = pd.read_csv(data_dir/'Friday-aft.csv')
fri_af2 = pd.read_csv(data_dir/'Friday-aft2.csv')

mon.head()
```

```python
# df.info()
```

```python
filename = [mon, tue, wed, th_m, th_af, fri_m, fri_af, fri_af2]
```

```python
for file in filename:
    file.columns = file.columns.str.strip()
```

```python
for i, file in enumerate(filename):
    print(f" {i} : {file['Label'].value_counts()}")
```

```python
df = pd.concat(filename, ignore_index = True)

df.head()
```

```python
# df.info()
```

```python
df['Label'] = df['Label'].str.strip()
```

```python
df['Label'].value_counts()
```

```python
label_class = selected_classes = [
    'BENIGN',
    'FTP-Patator',
    'SSH-Patator',
    'DoS Hulk',
    'DoS GoldenEye',
    'DoS slowloris',
    'DoS Slowhttptest',
    'DDoS',
    'PortScan'
]

filtered_data = df[df['Label'].isin(selected_classes)]
```

```python
merged_data = (filtered_data.groupby('Label', group_keys=False).sample(n=5000, random_state=42))
```

```python
merged_data.groupby("Label")['Label'].count()
```

```python
merged_data.to_csv(data_dir/'.csv')
```
